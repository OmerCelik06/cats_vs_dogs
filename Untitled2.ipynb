```python
import numpy as np
import tensorflow as tf
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow_datasets as tfds
from keras import layers
from keras.models import Sequential
from keras.losses import BinaryCrossentropy
from keras.optimizers import Adam
from keras.callbacks import EarlyStopping, ReduceLROnPlateau
from keras.preprocessing import image
from google.colab import files
```

```python
train_dataset = tfds.load('cats_vs_dogs', split='train[:80%]', as_supervised=True)
val_dataset = tfds.load('cats_vs_dogs', split='train[80%:]', as_supervised=True)
```

```python
def format_image(image, label):
  image = tf.image.resize(image, (224,224))
  image = image / 255.0
  return image, label
```

```python
train_dataset = train_dataset.map(format_image).batch(32).prefetch(tf.data.AUTOTUNE)
val_dataset = val_dataset.map(format_image).batch(32).prefetch(tf.data.AUTOTUNE)
```

```python

model = Sequential(
    [
    layers.Input(shape=(224, 224, 3)),

    # 1. Veri Çoğaltma (Hafifletilmiş - Zaman kazandırır)
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.05),

    # --- BLOK 1: Giriş (Burada standart Conv2D kullanmak en sağlıklısıdır) ---
    layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    # --- BLOK 2: Ayrılabilir Evrişim Başlıyor (Hızlanma Noktası) ---
    layers.SeparableConv2D(64, (3,3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    # --- BLOK 3: Orta Seviye Özellikler ---
    layers.SeparableConv2D(128, (3,3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    # --- BLOK 4: Mikro Detaylar (512 yerine 256 kullanıyoruz ki 3 saati aşmayalım) ---
    layers.SeparableConv2D(256, (3,3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    # --- KARAR MEKANİZMASI ---
    layers.GlobalAveragePooling2D(), # Parametre yükünü bitiren sihirli katman

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(1, activation='sigmoid')

    ]
)
lr_reducer = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6)

early_stopping = EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True)
```

```python
model.compile(
    optimizer = Adam(0.001),
    loss = BinaryCrossentropy(),
    metrics = ['accuracy']
)
model.summary()
```

```python
history = model.fit(
    train_dataset,
    validation_data = val_dataset,
    epochs = 40,
    callbacks = [early_stopping, lr_reducer]
)
```

```python
model.save('cats_vs_dogs.keras')
```
